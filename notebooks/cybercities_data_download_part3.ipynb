# Cybercities: part 3 (Cleveland business data)

Click **Runtime → Run all**, then upload `cbp_cleveland.csv` back to Claude.

```
import requests, pandas as pd
CENSUS_KEY = "YOUR_CENSUS_API_KEY"  # personal key: remove before publishing
url = ("https://api.census.gov/data/2023/cbp?get=NAME,NAICS2017,NAICS2017_LABEL,ESTAB,EMP"
       "&for=metropolitan%20statistical%20area/micropolitan%20statistical%20area:17410&key=" + CENSUS_KEY)
r = requests.get(url, timeout=120); r.raise_for_status(); rows = r.json()
df = pd.DataFrame(rows[1:], columns=rows[0])
df.to_csv("cbp_cleveland.csv", index=False)
print(len(df), "rows")
from google.colab import files
files.download("cbp_cleveland.csv")
```
